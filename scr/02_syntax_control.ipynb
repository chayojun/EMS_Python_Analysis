```python
# 기본함수 조차도 사용 X
nums3 = [5, 2, 8, 1, "abc"]
nums3_sum_val = 0
len_val = 0

for n in nums3:
    if str(n).isalpha() == False:
        len_val = len_val + 1
        nums3_sum_val = nums3_sum_val + n
nums3_sum_val / len_val
```

```python
# 숫자 범위를 벗어난 점수는 평균 계산에서 제외해야 한다.
score = [59, 92, 88, 0, 100, 1000, -77]
sum_val = 0
len_val = 0

for s in score:
    if s >= 0 and s <= 100:
        len_val = len_val + 1
        sum_val = sum_val + s
print(len_val)
print(sum_val / len_val)
```

## 1. 데이터 분석 실습

### 1. 데이터 준비(수집)

```python
students = [
    {"이름": "김철수", "성적": 95, "주민번호": "980101-1234567"},
    {"이름": "이영희", "성적": 88, "주민번호": "010305-4234567"},
    {"이름": "박민수", "성적": None, "주민번호": "950720-1234567"},  # 결측치
    {"이름": "최지은", "성적": 77, "주민번호": "030215-4234567"},
    {"이름": "정우성", "성적": 105, "주민번호": "990101-1234567"},   # 이상치
    {"이름": "한지민", "성적": 82, "주민번호": "020912-4234567"},
    {"이름": "강호동", "성적": -10, "주민번호": "700511-1234567"},   # 이상치
    {"이름": "신세경", "성적": 91, "주민번호": "971224-2234567"},
    {"이름": "유재석", "성적": 85, "주민번호": "720814-1234567"},
    {"이름": "송혜교", "성적": None, "주민번호": "811122-2234567"}, # 결측치
    {"이름": "차은우", "성적": 79, "주민번호": "970330-1234567"},
]

students
```

### 2. 결측치 확인

```python
missing_count = 0

for student in students: # 0 ~ 10 11번 돈다.
    if student["성적"] == None:
        missing_count = missing_count + 1

print("결측치 개수:", missing_count)
```

### 3. 결측치 처리(평균값 대체 : 어떤 값으로 대처하는지는 요구사항에 따라 다름)

```python
# 이상치를 처리하고 평균을 구해서 결측치 값을 구해야 했는데 안 했음.

scores = []

for student in students:
    if student["성적"] is not None: # !=
        scores.append(student["성적"])

print(scores)

avg = sum(scores) / len(scores) # 이상치 먼저 처리 수정필요

print(avg)

for student in students:
    if student["성적"] is None: # ==
        student["성적"] = avg

students # 쥬피트 노트북에서는 print(students)로 출력하는것보다 변수이름으로 출력하는 것이 출력결과의 가독성이 좋음
```

```python
for student in students:
    score = student["성적"]

    if score < 0 or score > 100:
        print(student)
```

### 7. 성별 구하기

```python
for student in students:

    # 문자.split(문자) : 하나의 잘라서 여러개의 문자 리스트 만들때 사용
    # 970330-1234567.split("-") ---> ["970330", "1234567"]
    # gender_code = "970330-1234567"[7:8]
    gender_code = student["주민번호"].split("-")[1][0]

    if gender_code in ["1", "3"]: # gender_code == 1 or gender_code == 3 
        student["성별"] = "남"
    else:
        student["성별"] = "여"

students
```

### 8. 나이 구하기(특성 추출)

```python
from datetime import datetime

current_year = datetime.now().year
# current_year = datetime.now().month, 일까지 추출해서 생일 지났는지 안 지났는지 확이해야함

print("current_year", current_year)

for student in students:
    rrn = student["주민번호"]
    yy = int(rrn[:2])
    code = rrn.split("-")[1][0]

    if code in ["1", "2"]:
        birth_year = 1900 + yy
    else:
        birth_year = 2000 + yy

    student["나이"] = current_year - birth_year

students
```

### 9. 성별 평균 성적

```python
male_sum = 0
male_count = 0

female_sum = 0
female_count = 0

for student in students:

    if student["성별"] == "남":
        male_sum += student["성적"]
        male_count += 1

    else:
        female_sum += student["성적"]
        female_count += 1

print("남 평균:", male_sum / male_count)
print("여 평균:", female_sum / female_count)
```

### 10. 연령대별 평균 성적

```python
age_group = {} # 빈 사전(dict) {"10대" : [2, 4, 6], "20대" : [3, 5, 7], "30대" : [10, 20, 30]}

for student in students:

    decade = (student["나이"] // 10) * 10  # 10, 20, 30, ....

    if decade not in age_group:  # key 값이 10, 20, 30 .... 없으면
        age_group[decade] = []

    age_group[decade].append(student["성적"])

age_group
```

```python
print(age_group)
for decade, scores in age_group.items():
    avg = sum(scores) / len(scores)
    print(f"{decade}대 평균: {avg:.2f}")
```

## 6. in, not in 연산자

* in 연산자는 어떤 값이 포함되어 있는지 확인할 때 사용합니다. -> True / False 반환

```python
nums = [10, 20, 30]

print(20 in nums) # True
text = "Python"

print("Py" in text) # True
student = {
    "이름": "김철수",
    "성적": 95
}

print("이름" in student) # True 딕셔너리에서는 키(key) 를 검사
fruit = "apple"

if fruit in ["apple", "banana", "orange"]:
    print("과일입니다.")
```

* not in 연산자 어떤 값이  포함되지 않았는지 검사 -> 포함되어 있지 않으면 True, 있으면 False 반환

```python
nums = [10, 20, 30]

print(40 not in nums) # True
```

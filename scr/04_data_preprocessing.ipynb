# 3. 람다함수 문법

* lambda 매개변수: 반환값

```python
# 일반함수
def add1(a, b):
    return a + b

r1 = add1(10,20)
r1
```

```python
# 람다함수 (익명함수, 일회용함수)
add2 = lambda a, b : a + b

r2 = add2(10,20)
r2
```

```python
def make(n, func): # n: 숫자, func: 함수
    return func(n) # 숫자
```

```python
# make 2번째 매개값으로 입력할 함수를 생성 (숫자, 함수): 숫자

def f1(n):
    return n*n  

def f2(n):
    return n*2

def f3(n):
    return "짝수" if n%2==0 else "홀수"

def f4(n):
    return n+n
```

```python
r = make(10, f2)
print(r)
```

```python
# f4 람다 형태로 변형
r = make(7, lambda n : n+n)
print(r)
```

```python
# f3 람다 형태로 변형
r = make(11, lambda x : "짝수" if x%2==0 else "홀수")
print(r)
```

```python
# f2 람다 형태로 변형
r = make(7, lambda n : n*2)
print(r)
```

```python
# f1 람다 형태로 변형
r = make(7, lambda n : n*n)
print(r)
```

# 5. 컬럼 이름을 사용하여 그룹화


- groupby()함수를 사용하여 그룹화 후 통계함수를 사용


- 예시) 성별 두개의 데이터셋을 분리 후 숫자컬럼값들의 평균을 구하고 두개의 집계결과를 하나로 병합하여(2행)하나의 DataFrame으로 반환


- 통계함수 사용시 컬럼명이 하나이면 ["컬럼명"], 둘이상이면 리스트를 사용하여 [["컬럼명1","컬럼명2"]]


- mean()

```python
import pandas as pd

df = pd.read_csv("../dataset/student_score.csv")
df.head()
```

```python
df2 = df.groupby("gender")
df2[["mid_score", "final_score"]].mean()
df2
```

```python
df2[["mid_score", "final_score"]].quantile(0.25)
```

```python
# 누적통계함수
df2["mid_score"].cumsum().head()
```

```python
# sum, min, max, mean, median 같은 이미 판다스에 내장된 집계 함수 중  인자가 없는(입력값이 없는) 함수는 
# 람다를 사용하지 않고도 사용 가능 
# 인자가 있는 함수나 사용자가 새로 정의한 집계함수는 람다식을 사용

df2["final_score"].agg(
    sum="sum",
    min="min",
    max="max",
    mean="mean",
    median="median"
)
```

```python
# []를 이용하여 간단하게 호출도 가능
df2["final_score"].agg(["sum","min","max","mean","median"])
```

```python
# x는 매개변수로 self( df2["final_score"] ) 자기 자신을 받는다.

df2["final_score"].agg([
    lambda x: x.quantile(0.25),  # x --> df2["final_score"]
    lambda x: x.quantile(0.50),
    lambda x: x.quantile(0.75)
])
```

## 5. boxplot

​
언제?

- 이상치 탐지 

- 그룹 비교

​
분석

- 이상치가 있는가?

예시) 2000이 다른 값들과 너무 멀어 이상치로 표시됨.


- 박스플롯은 다음 정보를 한 번에 보여줍니다.

최소값, Q1 (25%), 중앙값(Q2), Q3 (75%), 최대값, 이상치(최소/최대는 이상치를 제외한 값들 중)

```python
import numpy as np

salary = [
    280, 290, 300, 310, 320,
    330, 340, 350, 360, 370,
    380, 390, 400, 410, 420,
    430, 440, 450, 460, 2000
]

# 급여 평균을 구하면

np.mean(salary) # 451.5 인데 대부분은 450 아래 -> 평균만 보면 착시 발생
```

```python
import matplotlib.pyplot as plt

plt.boxplot(salary)

plt.ylabel("salary")
plt.show()


# 
#
#       ●  ← 이상치 발견(2000)
# 
#       
#      ---- 최대값(이상치 제외 후)
#       |
#  ┌───────┐ 제3사분위수
#  │---■---│  (중앙값) 제2사분위수
#  └───────┘ 제1사분위수
#       |
#      ---- 최소값(이상치 제외 후)


# Q1 = 320만원, 중앙값 = 380만원, Q3 = 430만원
# 박스플롯을 통해 "2000만원의 이상치가 평균을 왜곡하고 있다"는 사실을 발견할 수 있으며, 
# 이후 이상치 제거·분리 분석·중앙값 사용 등의 의사결정을 할 수 있습니다.
# 박스플롯은 데이터의 분포(중앙값, 퍼짐 정도)와 이상치를 한눈에 확인하여 데이터 품질을 점검하기 위해 사용한다.
```

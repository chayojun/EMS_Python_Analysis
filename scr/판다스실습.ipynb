```python
# 기술통계 분석
"""
count (개수)뜻: 시험을 본 학생 수 (데이터 개수)설명: 국어 시험을 총 몇 명이 쳤는지 보여줍니다. 만약 이 숫자가 다른 과목보다 적다면 시험을 안 본(결석한) 학생이 있다는 뜻입니다.
mean (평균)뜻: 우리가 잘 아는 평균 점수설명: 점수를 다 더한 뒤 학생 수로 나눈 값입니다. 반 평균이 얼마인지 보여줍니다.
std (표준편차)뜻: 점수들이 평균에서 얼마나 멀리 떨어져 있는지(퍼져 있는지) 나타내는 지표설명: 이 값이 작으면 학생들 점수가 평균 주변에 다닥다닥 모여 있는 것(실력이 비슷함)이고, 크면 잘하는 학생과 못하는 학생의 점수 차이가 크다는 뜻입니다.
min (최솟값)뜻: 꼴등 점수 (가장 낮은 점수)
25% (제1사분위수)뜻: 학생들을 점수 낮은 순서부터 일렬로 세웠을 때, 앞에서 25%(하위 1/4) 지점에 있는 학생의 점수설명: 뒤에서 등수를 셌을 때 이 점수보다 낮으면 공부를 조금 더 열심히 해야 하는 구간입니다.
50% (중앙값)뜻: 정확히 가운데(50%)에 서 있는 학생의 점수설명: 만약 반에 딱 한 명만 유독 0점을 맞아서 평균(mean)이 깎였더라도, 이 중앙값은 영향을 받지 않고 진짜 중간 실력을 보여줍니다.
75% (제3사분위수)뜻: 점수 낮은 순서로 세웠을 때 75% 지점(상위 1/4)에 있는 학생의 점수설명: 내 점수가 이 점수보다 높다면 반에서 상위권(잘하는 편)에 속합니다.
max (최댓값)뜻: 1등 점수 (가장 높은 점수)
"""

df.describe(include='all')
```

```python
# 4) 전치러
df = df.replace(r'^\s*$', None, regex=True) # 빈문자열, 공백 등을 None 으로 변경
df
```

```python
"""
- NaN 숫자자료형 변수의 결측시, None은 그외 자료형의 결측시 
- 문자형 이상값은 NaN 처리
- 'A', '결시', '@', None, '', .. 값들을 NaN으로 변경
- NaN의 연산: NaN은 숫자로 취급되므로 수학 연산을 해도 에러가 나지 않고 결과가 NaN이 됩니다. 
  (예: NaN + 5 = NaN) 또한 판다스의 합계(.sum())나 평균(.mean()) 연산 시 자동으로 무시(제외)하고 계산됩니다.
- None의 연산: 파이썬 객체이기 때문에 일반적인 숫자나 문자열과 더하려고 하면 타입 에러(TypeError)가 발생합니다. 
  다만, 판다스 함수 내부에서는 None도 결측치로 인식하여 안전하게 처리해 줍니다. -> 그래도 숫자값에 None은 NaN으로 변경한는게 안
"""
score_cols = ['kor','eng','math'] # 전처리 계산에 사용할 숫자 컬럼 리스트

for col in score_cols: # col값은 'kor', 'eng', 'math' 3번 반복
    # 글자(문자열)로 되어 있는 컬럼을 숫자 데이터로 바꾸되, 숫자로 바꿀 수 없는 글자는 에러를 말고 결측치(NaN)로 만들어
    # ('30' -> 30), ('abc' -> NaN), (None -> NaN)

    df[col] = pd.to_numeric(df[col], errors='coerce') 
    # df['kor'] 값을 전부다 숫자로 변경해라, # df['eng'] 값을 전부다 숫자로 변경해라, # df['math'] 값을 전부다 숫자로 변경해라

df.head()
```

```python
# 이상치 처리 
for col in score_cols:
    df.loc[(df[col] < 0) | (df[col] > 100), col] = np.nan

df.head()
```

```python
# 이상값 수정 후 다시 결측치 확인
df.isnull().sum()
```

```python
# 결측치 평균값으로 대체
for col in score_cols:
    a = df[col].mean()
    df[col] = df[col].fillna(a) 
    # df[col].mean() 평균을 구해서 그 평균값을 df[col].fillna(평균값)
    # fillna의 반환값은 float 타입이라서 소수점으로 바뀌었다.
df.head()
```

```python
# 5) 데이터 가공
# 총점 파생변수(컬럼) 추가
df['total'] = df['kor'] + df['eng'] + df['math']
df.head()
```

```python
# 평균 파생변수 추가
df['avg'] = df['total'] / 3
df.head()
```

```python
# 등급 파생변수 추가
def grade(avg):
    if avg >= 90:
        return 'A'
    elif avg >= 80:
        return 'B'
    elif avg >= 70:
        return 'C'
    else:
        return 'D'

df['grade'] = df['avg'].apply(grade) 
# apply 함수는 판다스 문법 수업에서 설명(파이썬 함수의 이해가 필)

df.head()
```

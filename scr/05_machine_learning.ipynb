# 1. 인공지능, 머신러닝, 딥러닝

1. 인공지능 이란?

​

- 1차 인공지능 붐 : 사전에 입력된 코드를 통해 계산을 하는 인공지능. 준비된 코드의 계산만 가능.

​

- 2차 인공지능 붐: 전문 지식을 데이터로 입력해 두고 사용자의 질문에 대답할 수 있게 만듬. 많은 전문 지식과 사용자 질문 규칙등을 만들어야 해서 엄청난 비용(노력, 시간, ...)을 초래함.

​

- 3차 인공지능 붐(현재) : 인터넷등의 환경을 통해 대량의 데이터를 활용해서, 컴퓨터(기계) 스스로 학습해 가는 머신러닝(기계학습)을 개발함. 머신러닝은 사용자가 데이터를 주고 이 데이터들의 어떤 특징에 주목하여 학습하라고 하면 기계스스로 학습하는 방법.

​

* 딥러닝(심청학습) : 머신러닝의 한 방법으로 "어떤 특징에 주목하여 학습하라"고 지시하지 않아도, 기계가 데이터에서 스스로 특징을 찾아내는 방법이다. 오늘날의 인공지능이라고 하면 대부분 "딥러닝"을 가리킨다.

​

​

2. 머신러닝

​

- 데이터가 가지는 특징의 법칙성을 학습시키는 것.

- 데이터 속에서 중요한 특징을 찾아내고, 그 특징을 학습시킬 "모델"을 준비하여, 준비된 모델에 특징을 학습시켜 "학습된 모델"만든다.

- 이렇게 "학습된 모델"을 이용하여 새로운 경우에 대해서 결과를 예측한다.

ex) 

여러 꽃 데이터에서 꼭에 대한 특징을 학습한 모델을 생성하여, 어떤 꽃의 데이터를 가지고 이 꽃이 무엇인지 예측한다.

과거 매출데이터를 학습하여 학습된 모델을 만들고, 올해 매출을 예측한다.

​

- 데이터가 적으면 편향된 학습이 되고, 품질을 떨어지는 데이터는 엉뚱한 학습 모델을 만든다. 그래서 머신러닝에서 데이터의 양이나 품질이 중요하다. 데이터 수집이나 데이터 분석이 중요한 이유다.



* 머신러닝의 종류

지도학습 : 수치나 분류를 예측하는 학습

비지도학습 : 데이터를 정리(클러스터링)하는 학습

강화학습 : 경험을 통해 능숙해지는 학습

​

​

1. 지도학습

​

- 데이터(이미지, 문자, 등..)에 문제/답형태 준비된 "훈련 데이터"를 사용(지도)하여 학습하는 방법

- 회귀 : 어떤 값과 관련된 값이 어떤 수치가 될 지 예측하는 지도학습방법

- 분류 : 어떤 데이터가 어느 분류에 들어가는지 예측하는 지도학습방법

ex) 

-사람의 사진 데이터 -> 남자/여자를 표기(훈련데이터) -> 새로운 사진의 남자/여자를 예측

- 훈련데이터로 한국어(한글 자음, 모음), 영어(대문자,소문자), 일본어(히라가나, 가타카나) 학습 후 단어를 영어/일본어/한국어로 분류 

​

2. 비지도학습

​

- 문제의 형태만으로 훈련데이터없이(비지도) 학습하는 방법이다. 답을 찾는 것이 아니라 데이터를 분류(클러스터링)하고 복잡한 데이터를 간결하게(차원축소) 한다.

​

- 클러스터링 : 많은 데이터를 그룹으로 나눌때 사용하는 비지도학습방법, 비슷한 특성을 가진 데이터끼리 자동으로 그룹을 만드는 비지도학습 기법

​

- 사람의 사진데이터 -> 두개의 그룹(남/여)으로 데이터를 나눈다. 

  주의) 지도학습 처럼 남/여 타겟을 정해놓고 분류하는 것(지도학습의 분류)이 아니고(y 가없다. x만으로 분류) 

         주어진 데이터를(분석) 설명변수를 통해 분석해보니 두 그룹정도로 나누는게(클러스터링) 좋겠다. 나누고 보니            남/여로 나누어짐. 

         -> 분석 방법에 따라 3그룹(남/여/유아), 4그룹(남유아/남성인/여유아/여성인)으로 나누어 지기도 함.

         그래서 새로운 행이 입력되면 두 그룹중 하나로 평가하는 방법 

​

- 많은 데이터(한국어, 영어, 일본어 단어)들을  3분류로 클러스터링

 

​

3. 강화학습

​

여러가지 방법으로 학습하게 하고, 결과에 따라서 상/별을 주는 방법

단 하나의 답을 찾는 것이 아니라 더 나은 방법을 찾아내기 위한 방법이다.

지도학습은 정답을 맞추는 것이 목표이고, 강화학습은 결과적으로 보상을 최대화하는 행동을 배우는 것이 목표입니다.

단순 주식같은 예제에서는 둘이 매우 비슷해 보일 수 있습니다.

​

- ex)

- 장기, 바둑, 게임, AI, 자율주행

- 자율주행차가 도로에 학습시키는 경우

차선 유지 → +10

차선 이탈 → -20

앞차와 충돌 → -100

목적지 도착 → +100

이런 보상 시스템으로 안전하게 운전하도록 학습시킨다.

## 1. 머신러닝 맛보기1(지도학습 분류 iris 데이터셋)

```python
# scikit-learn의 샘플 데이터 불러오기

from sklearn import datasets
iris = datasets.load_iris() # 붓꼿을 세가지 품종으로 분류하는 학습을 할 수 있는 데이터 세트

iris
```

```python
print("특징량 이름=", iris.feature_names)
# 특성(피처)
# x1 'sepal length (cm)' 꽃받침 길이
# x2 'sepal width (cm)'       폭
# x3 'petal length (cm)' 꽃잎 길이
# x4 'petal width (cm)'       폭
```

```python
print("분류(품종) 이름=", iris.target_names) 
# 'setosa' : 알래스카 훗카이도 분포 
# 'versicolor' : 미국동부 캐나다 동부에 분포 
# 'virginica' : 미국남동부에 분포
```

```python
# y 타겟(정답)
print("분류 값=", iris.target)
```

```python
import pandas as pd
df = pd.DataFrame(iris.data)
df.head() # 앞에서 다섯줄 표시
```

```python
# 열 이름[0,1,2,3]을 ['sepal length (cm)', 'sepal width (cm)', 'petal length (cm)', 'petal width (cm)'] 으로 변경
df.columns = iris.feature_names 
df
```

```python
# "target"열을 추가하고 품종 분류값인 iris.target리스트 값을 입력 
df["target"] = iris.target
df.head(5)
df.head(-5)
```

```python
# Jupyter Notebook에서 그래프를 셀 바로 아래에 출력하라는 IPython 매직 명령어
%matplotlib inline 
import matplotlib.pyplot as plt
```

```python
# 3그룹
# 1. 학습 데이터(train)와 평가 데이터(test)로 분리
from sklearn.model_selection import train_test_split


# 위(3~4셀)에 만든 df를 X, y 데이터 프레임으로 분리
# X : 입력 데이터(꽃받침/꽃잎 길이, 너비)
# y : 정답 데이터(꽃 종류)
X = df[
    [
        'sepal length (cm)',
        'sepal width (cm)',
        'petal length (cm)',
        'petal width (cm)'
    ]
]

y = df["target"]


# Iris 데이터셋은 학습을 위한 쉬운 데이터라서 여러 모델이 거의 100% 정확도를 보입니다.
# 학습을 위해 학습데이터를 줄이고, 테스트 데이터를 늘려서 아래 알고리즘별 정확도 비교(test_size=0.5 아래로 해서)

X_train, X_test, y_train, y_test = train_test_split(
    X,              # 입력 데이터
    y,              # 정답 데이터
    test_size=0.5,  # 전체의 20%를 테스트용으로 사용 
    random_state=42 # 항상 같은 결과가 나오도록 난수 고정
)

# 결과 예시
# X_train : 120개
# X_test  : 30개
# y_train : 120개
# y_test  : 30개
```

```python
print(X_train)
```

```python
print(y_train)
```

```python
print(X_test)
```

```python
print(y_test)
```

```python
# 머신러닝 모델 생성
from sklearn.neighbors import KNeighborsClassifier

# 어떤 알고리즘 모델을 선택할것인가는 데이터따라 달라진다. 
# 머신러닝 알고리즘을 공부하는 이유가 이러한 데이터에는 어떤 알고리즘을 사용하는 것이.... 
# KNN(최근접 이웃 알고리즘) 알고리즘을 가진 모델을 생성
model = KNeighborsClassifier()
# 머신러닝 학습
# X_train값을 입력하고 y_train값을 확인하면서 어떤 입력에 대해서 어떤 y값이 나오는지 규칙을 KNN알고리즘을 이용하는 모델로 학습
# ex) 꽃잎 길이가 짧으면 setosa, 꽃잎 길이가 길면 virginica
model.fit(X_train, y_train) # 120 X_train -> 120 y_train

# 결과 이미지 해석
# | 파라미터          | 값         | 의미                    |
# | ------------- | --------- | --------------------- |
# | n_neighbors   | 5         | 가장 가까운 이웃 5개를 참고하여 분류 |
# | weights       | uniform   | 5개 이웃 모두 동일한 영향력      |
# | algorithm     | auto      | 최적의 탐색 알고리즘 자동 선택     |
# | leaf_size     | 30        | 탐색 트리의 크기 설정          |
# | p             | 2         | 거리 계산 시 유클리드 거리 사용    |
# | metric        | minkowski | Minkowski 거리 사용       |
# | metric_params | None      | 거리 계산 추가 옵션 없음        |
# | n_jobs        | None      | CPU 기본 설정 사용          |
```

```python
# 학습된 모델을 이용하여 남겨둔 테스트용 데이터 30개(X_test -> y_test) 얼마나 정확한지 예측
# 4. 학습된 모델로 예측
pred = model.predict(X_test)
pred
```
